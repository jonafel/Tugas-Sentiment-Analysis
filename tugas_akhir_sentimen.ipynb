```python
def text_cleaning(text:str):
  text_cleaned = text.lower()

  pattern = re.compile(r'@\w+')
  clean_url_text = pattern.sub("", text_cleaned)

  clean_number_text = re.sub(r"[\d]+.[\d]+|[\d]+","", clean_url_text)

  pattern = re.compile(r'\(?https?://[^\s$#]+\)?')
  clean_url_text = pattern.sub("", clean_number_text)


  clean_punct_text = clean_url_text.translate(str.maketrans('','',string.punctuation))

  clean_wspace_text = clean_punct_text.strip()
  clean_wspace_text = clean_wspace_text.replace('\n','')
  clean_wspace_text = re.sub(r"[\s][.,]?[\s]"," ",clean_wspace_text)

  clean_wspace_text = re.sub(r'\b(\w+)\1\b', r'\1', clean_wspace_text)

  return clean_wspace_text
```

[StopWord Bahasa Indonesia](https://drive.google.com/file/d/1E844kUnMSCguYSQcNZzj62aJjQX1HwVO/view?usp=drive_link)

```python
def load_stopwords(filepath):
      try:
          df_stopword = pd.read_csv('stopwordbahasa.csv', encoding='utf-8')
          stopword_list = df_stopword.iloc[:,0].tolist()
          stopword_set = set(stopword_list)

          return stopword_set
      except FileNotFoundError:
          print(f"File not found: {filepath}")
          return set()
```

```python
def stopword_cleaning(text, stopword_set):
        text_words = text.split()
        cleaned_words = [word for word in text_words if word not in stopword_set]
        return ' '.join(cleaned_words)
```

```python
stopword_sets = load_stopwords('stopwordbahasa.csv')
stopword_sets
```

## METODE NAIVE BAYES

[Kata Positif](https://drive.google.com/file/d/1YW3fqnBDeBa6-UF5HjZWCjqbENOwdSqp/view?usp=drive_link)
[Kata Negatif](https://drive.google.com/file/d/1mJkAtp0rWRSZnpvH0JwC27fZCKvF_UXP/view?usp=drive_link)

```python
with open("./kata positif.txt","r") as positif_file :
  positive_words = list(set(positif_file.readlines()))

with open("./kata negatif.txt","r") as negative_file :
  negative_words = list(set(negative_file.readlines()))
```

```python
positive_words = [word.rstrip() for word in positive_words]
negative_words = [word.rstrip() for word in negative_words]

print(positive_words[:10])
print(negative_words[:10])
```

```python
def sentiment_counts(text, positive_words, negative_words):
  good_count = 0
  bad_count = 0

  tokens = word_tokenize(text)

  for token in tokens:
        if token in positive_words:
            good_count += 1
        if token in negative_words:
            bad_count += 1

  return good_count - bad_count
```

```python
data['count'] = data['text_non_slang'].apply(lambda x: sentiment_counts(x, positive_words, negative_words))
```

```python
data
```

pada kolom count, bisa kita lihat banyak angka yang menunjukkan kearah mana kalimat di depannya positif atau negatif. jadi kita buat kolom sendiri yang dapat memberikan hasil sentimennya.

```python
def sentimen(score):
  if score > 0:
    return "positif"
  elif score == 0:
    return "netral"
  else:
    return "negatif"
```

```python
data['hasil sentimen'] = data['count'].apply(sentimen)
data
```

```python
data['hasil sentimen'].value_counts()
```

##METODE LEXICON

Kita coba data yang sama tapi dengam metode yang berbeda
[Positif Lexicon](https://drive.google.com/file/d/1JnwJW4aQCJTZ5N2gtSOb-Dmcu1Pnamt8/view?usp=drive_link)
[Negatif Lexicon](https://drive.google.com/file/d/1gQRv8cFpBDNSPy5MYUzo3d-Z4Aaswl_M/view?usp=drive_link)

```python
factory = StemmerFactory()
stemmer = factory.create_stemmer()

with open("./lex positif.txt","r") as positif_file :
  positive_words_lex = list(set(positif_file.readlines()))

with open("./lex negatif.txt","r") as negative_file :
  negative_words_lex = list(set(negative_file.readlines()))

positive_lex = [word.rstrip() for word in positive_words_lex]
negative_lex = [word.rstrip() for word in negative_words_lex]
```

```python
def sentiment_counts_lex(text, positive_lex, negative_lex, stemmer):

  tokens = word_tokenize(text)
  stemmed_tokens = [stemmer.stem(token) for token in tokens]

  positive_count = 0
  negative_count = 0

  for token in stemmed_tokens:
        if token in positive_lex:
            positive_count += 1
        if token in negative_lex:
            negative_count += 1

  sentiment_score = positive_count - negative_count

  if sentiment_score > 0:
        return "positive"
  elif sentiment_score == 0:
        return "neutral"
  else:
        return "negative"
```

```python
data['lexicon_sentiment'] = data['text_non_slang'].apply(lambda x: sentiment_counts_lex(x, positive_lex, negative_lex, stemmer))
data
```

```python
data['lexicon_sentiment'].value_counts()
```

Proses yang sama tapi hasil yang berbeda walaupun tetap komentar sentimen positif ada di nomor 1, diikuti oleh negatif dan netral. Hasil disini jauh lebih kompetitif
